```python
import pandas as pd
```

```python
train = pd.read_csv("../data/train.csv")
train.head()
```

```python
# 选择一列
# 假设只想查看乘客年龄
train["Age"]
```

```python
# 选择多列
train[["Name", "Sex", "Age"]]
```

```python
# loc: 根据 "名字" 选择
# iloc: 根据 “位置” 选择

# 第0行
train.loc[0]
```

```python
# 0-4行，包含4
train.loc[0:4]
```

```python
# 用loc同时选择行和列
train.loc[0:4, ["Name", "Sex", "Age"]]
```

```python
# 第一个位置的行
train.iloc[0]
```

```python
# 不包含末尾
train.iloc[0:5]
```

```python
# Titanic 中有哪些乘客年龄超过 60 岁
train["Age"] > 60
```

```python
train[train["Age"] > 60]
```

```python
# 找所有女性乘客
train[train["Sex"] == "female"]
```

```python
# 找所有幸存者
train[train["Survived"] == 1]
```

```python
# 找年龄超过 60 岁的乘客，但我只关心姓名、性别、年龄和是否存活
train.loc[train["Age"] > 60, ["Name", "Sex", "Age", "Survived"]]
```

```python
# 年龄 < 18 并且是女性的乘客有哪些
train[
    (train["Age"] < 18) &
    (train["Sex"] == "female")
]
```

```python
# 年龄 < 18 或者年龄 > 60
train[
    (train["Age"] < 18 |
    (train["Age"] > 60))
]
```

```python
# 一等舱或者二等舱
train[train["Pclass"].isin([1, 2])]
```

```python
# Titanic 年龄最大的乘客是谁
train.sort_values("Age")
train.sort_values("Age", ascending=False)
```

```python
train.sort_values(
    "Age",
    ascending=False
)[["Name", "Sex", "Age", "Survived"]].head(10)
```

```python
# 找票价最高的乘客
train.sort_values(
    "Fare", ascending=False
)[["Name", "Pclass", "Fare"]].head(10)
```

```python
# 先按照 Pclass 排序，如果舱位相同，再按照 Fare 从高到低
train.sort_values(
    ["Pclass", "Fare"], ascending=[True, False]
)
```

```python
# Titanic 里男性和女性分别有多少人
train["Sex"].value_counts()
```

```python
# 统计幸存人数
train["Survived"].value_counts()
```

```python
# 看比例而不是人数
train["Sex"].value_counts(normalize=True)
```

```python
# 连缺失值也一起统计
train["Embarked"].value_counts(dropna=False)
```

```python
# Sex 到底有哪些不同的值
train["Sex"].unique()
```

```python
# 一共有多少种不同的值
train["Sex"].nunique()
```

```python
# 整体幸存率大概是多少
train["Survived"].value_counts(normalize=True) * 100
```

```python
train["Pclass"].value_counts(normalize=True) * 100
```

```python
# 男性和女性的幸存率分别是多少
train.groupby("Sex")["Survived"].mean()
```

```python
# 不同舱位的幸存率
train.groupby("Pclass")["Survived"].mean()
```

```python
# 每个舱位平均票价
train.groupby("Pclass")["Fare"].mean()
```

```python
# 每个舱位：
# 平均票价
# 最低票价
# 最高票价
train.groupby("Pclass")["Fare"].agg(["mean", "min", "max"])
```

```python
# 两个字段一起分组
train.groupby(["Sex", "Pclass"])["Survived"].mean()
```
